# API Walkthrough


```python
from pathlib import Path
import shutil

import matplotlib.pyplot as plt
from PIL import Image

from slide2vec import ExecutionOptions, Model, PreprocessingConfig

SLIDE_PATH = "data/histai/wsi-fixed/slide_H&E_5.tif"
MASK_PATH = "data/histai/mask/slide_H&E_5.tif"
OUTPUT_DIR = Path("tutorials") / "output" / "api_walkthrough"

if OUTPUT_DIR.exists():
    shutil.rmtree(OUTPUT_DIR)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"slide: {SLIDE_PATH}")
print(f"mask: {MASK_PATH}")
print(f"output: {OUTPUT_DIR}")
```

## 1. Authenticate with Hugging Face

PRISM is a gated model. Request access on its Hugging Face page, then run `hf auth login` or set `HF_TOKEN` before you start the notebook.


## 2. Embed the slide with PRISM


```python
model = Model.from_preset("prism")
preprocessing = PreprocessingConfig(
    backend="asap",
    preview={
        "save_mask_preview": True,
        "save_tiling_preview": True,
        "downsample": 32,
    },
)
execution = ExecutionOptions(output_dir=OUTPUT_DIR, num_gpus=1)
```

```python
embedded = model.embed_slide(
    SLIDE_PATH,
    preprocessing=preprocessing,
    execution=execution,
    mask_path=MASK_PATH,
)

print(f"sample_id: {embedded.sample_id}")
print(f"tile_embeddings shape: {embedded.tile_embeddings.shape}")
print(f"slide_embedding shape: {embedded.slide_embedding.shape}")
```

```python
tissue_mask_preview = Image.open(embedded.mask_preview_path)
tissue_mask_preview.show()
```

```python
tiling_preview = Image.open(embedded.tiling_preview_path)
tiling_preview.show()
```
